# Lambda functions

```python
def hello(to='world'):
    print(f"hello, {to}")
```

```python
hello()
```

```python
hello('Adhi')
```

```python
h = lambda to='world':f"hello {to}"
```

```python
h('Adhi')
```

```python
h()
```

```python
def add(a,b,c):
    print(a+b+c)
add(1,2,3)
```

```python
add = lambda a,b,c:a+b+c
add(1,2,3)
```

# List Comprehension

```python
n = []
for i in range(10):
    n.append(i)
n
```

```python
[n for n in range(10)]
```

```python

age = int(input())
if age > 18:
    print(True) 
else:
    print(False)
```

```python
age = int(input())
print(True if age>18 else False)
```

```python
[n for n in range(1,20)]
```

```python
[n for n in range(1,20) if n%2!=0]
```

```python
[n**2 if n%2==0 else n for n in range(20)]
```

```python
[(lambda i:i**2)(n) if n%2==0 else n for n in range(10)]
```
